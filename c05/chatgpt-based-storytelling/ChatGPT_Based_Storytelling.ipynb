# ChatGPT-Based Storytelling
**Course End Project - Essentials of Generative AI, Prompt Engineering, and ChatGPT**

The brief asks for an interactive story built collaboratively with ChatGPT. Because a ChatGPT web session cannot be recorded from a notebook, every prompt below is sent programmatically and every response shown is a real, executed model reply.

> **OpenAI GPT-4o-mini accessed via OpenRouter's OpenAI-compatible API.** (API keys are loaded from a local `.env` at runtime and are never printed or saved.)

The conversation keeps a running message history, exactly like a ChatGPT chat window, so each reply builds on the previous turns.

```python
import os, json, copy
from llm import chat, MODEL
from common import render_text_png
os.makedirs('screenshots', exist_ok=True)
SYSTEM = ("You are the narrator and game master of an interactive text adventure. Write vivid but concise prose "
          "(max ~130 words per reply), stay consistent with established characters and facts, and end story turns "
          "by offering the player 2-3 numbered choices unless told to conclude.")
history = [{"role": "system", "content": SYSTEM}]
log = []   # (step, prompt, reply)
def turn(step, prompt, shot=None, max_tokens=320):
    history.append({"role": "user", "content": prompt})
    reply = chat(history, max_tokens=max_tokens, temperature=0.8)
    history.append({"role": "assistant", "content": reply})
    log.append((step, prompt, reply))
    print("USER >>", prompt, "\n"); print("CHATGPT >>", reply)
    if shot: render_text_png(step, [("You", prompt), ("ChatGPT (gpt-4o-mini)", reply)], f"screenshots/{shot}.png")
    return reply
print("Model:", MODEL, "via OpenRouter")
```

## Step 1 - Define the story theme
Genre chosen: **fantasy adventure with a mystery element** (a lighthouse at the edge of the world whose light has gone out).

```python
turn("Step 1 - Define the story theme",
     "Let's build an interactive fantasy-mystery adventure together. Theme: the last lighthouse at the edge of the world "
     "has gone dark, and the sea is starting to swallow nearby islands. Tone: wondrous but slightly eerie. "
     "Confirm the theme in 2 sentences and ask me nothing else yet.", "01_theme")
```

## Step 2 - Introduce the main characters
Characters are defined by name, personality and background so the model keeps them consistent.

```python
turn("Step 2 - Introduce the main characters",
     "Main characters:\n"
     "1) Mira Voss - me, the player. 19-year-old apprentice cartographer, curious and stubborn, grew up on the island of Tern, "
     "carries a brass compass that points to lost things.\n"
     "2) Captain Odric Hale - grizzled ferryman, sarcastic but loyal, once served the lighthouse keeper.\n"
     "3) Wick - a talking storm-petrel, the keeper's former messenger, nervous and secretly knows why the light failed.\n"
     "Briefly acknowledge each character in one line, keeping their personalities.", "02_characters")
```

## Step 3 - Begin the story
Set the scene and give the story a starting point.

```python
turn("Step 3 - Begin the story",
     "Begin the story. Open on Mira, Odric and Wick aboard Odric's ferry approaching the dark lighthouse at dawn. Set the scene "
     "and end with a first choice.", "03_begin")
```

## Step 4 & 5 - Participant input and AI-generated responses
The player describes the actions of their character; the model responds with narrative that expands the plot.

```python
turn("Step 4/5 - Participant input and AI response",
     "Mira raises her brass compass to see where it points, and quietly asks Wick what he is hiding. (I choose to investigate "
     "the compass and question Wick.)", "04_player_action")
```

```python
turn("Step 5 - Expand the story",
     "Odric docks the ferry. Mira steps onto the rocks. Describe what she finds at the lighthouse door and let the plot advance "
     "with one new clue. End with choices.", "05_expand")
```

## Step 6 - Decision-making and branching paths
I save the conversation state here and explore **two different branches** from the same choice to show how decisions lead to different outcomes.

```python
turn("Step 6 - Branching decision",
     "Offer me a decisive fork now: (A) climb the spiral stairs to the lamp room, or (B) descend into the flooded cellar. Give 2 sentences on the risk of each.", "06_decision")
branch_point = copy.deepcopy(history)
```

```python
print("=== BRANCH A: climb to the lamp room ===")
history[:] = copy.deepcopy(branch_point)
branch_A = turn("Step 6 - Branch A (lamp room)", "I choose A. Mira climbs to the lamp room. Narrate the discovery and its consequence in ~100 words.", "06a_branch_A")
stateA = copy.deepcopy(history)
```

```python
print("=== BRANCH B: descend into the cellar ===")
history[:] = copy.deepcopy(branch_point)
branch_B = turn("Step 6 - Branch B (cellar)", "I choose B. Mira descends into the flooded cellar. Narrate the discovery and its consequence in ~100 words.", "06b_branch_B")
```

**Interpretation:** the two branches start from the identical state but lead to different discoveries, showing how player choices steer the narrative. I continue with Branch A as the canonical story.

```python
history[:] = copy.deepcopy(stateA)
```

## Step 7 - Iteration and refinement
I refine my input for clarity and consistency: I remind the model of character traits it should respect and ask for a rewrite of the last scene.

```python
turn("Step 7 - Iteration and refinement",
     "Refine that: Captain Odric must stay sarcastic and Wick must remain nervous, and the brass compass must play a role. "
     "Rewrite the last scene (~100 words) keeping all facts the same but fixing the tone, then give me 2 choices.", "07_refine")
```

```python
turn("Step 7 - Continue the story",
     "I pick the bolder choice. Mira trusts the compass and acts. Advance the plot toward the climax (~100 words), with one final choice.", "07b_continue")
```

## Step 8 - Conclusion
I guide the story towards a satisfying ending. Decision: the adventure ends with a **resolved ending plus a hook** for a possible sequel.

```python
turn("Step 8 - Conclusion",
     "Now conclude the story in ~150 words: resolve the mystery of the dark light, give Mira, Odric and Wick each a closing beat, "
     "and end with a one-line open hook for a possible sequel. No more choices.", "08_conclusion", max_tokens=380)
```

## Step 9 - Showcase and share the story
The canonical path (opening, action, expansion, Branch A, refined scene, continuation, conclusion) is assembled from the narrator's replies and saved as a shareable text file.

```python
keys = ["Step 3 ", "Step 4/5", "Step 5 ", "Branch A", "Step 7 - Iteration", "Step 7 - Continue", "Step 8"]
canon = [r for (s, p, r) in log if any(k in s for k in keys)]
text = "\n\n".join(canon)
open("The_Lighthouse_at_the_Edge_of_the_World.txt", "w", encoding="utf8").write(text)
print(text)
render_text_png("Final story (canonical path)", [("The Last Lighthouse", text)], "screenshots/09_final_story.png")
print("\nTurns:", len(log), "| words in story:", len(text.split()))
```
